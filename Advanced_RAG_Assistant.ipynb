```python
def ask_llm(question):

    message = [
        {
            "role": "system",
            "content": "You are a helpful AI assistant. Answer clearly."
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt = tokenizer.apply_chat_template(
        message,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        do_sample=False,
        temperature=0.7,
        top_k=4
    )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response


answer = ask_llm("What is Retrieval Augmented Generation?")
print(answer)
```

```python

def ask_llm(question):

    messages = [
        {
            "role": "system",
            "content": "You are a helpful AI assistant. Answer clearly."
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=False
    )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response.strip()


answer = ask_llm("Explain Retrieval Augmented Generation in simple words.")
print(answer)


```

```python
import re

def clean_text(text):
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    text = text.strip()
    return text

for doc in documents:
    doc.page_content = clean_text(doc.page_content)

print("Document cleaning completed!")
print(documents[0].page_content[:1500])
```

```python

from typing import TypedDict
from langchain_core.documents import Document

class RAGState(TypedDict):

    question: str
    documents: list[Document]
    answer: str
```

```python

def retrieve(state: RAGState):

    question = state["question"]

    documents = loaded_vectorstore.similarity_search(
        question,
        k=3
    )

    return {"documents": documents}
```

```python

def generate(state: RAGState):

    question = state["question"]

    context = "\n\n".join(
        doc.page_content for doc in state["documents"]
    )

    prompt = f"""
You are a helpful assistant.
Answer the question using ONLY the provided context.

If the answer is not available in the context,
clearly say that the information is not available.

Context:
{context}

Question:
{question}

Answer:
"""

    answer = ask_llm(prompt)

    return {"answer": answer}
```

```python

from langgraph.graph import StateGraph, START, END

graph_builder = StateGraph(RAGState)

graph_builder.add_node("retrieve", retrieve)
graph_builder.add_node("generate", generate)

graph_builder.add_edge(START, "retrieve")
graph_builder.add_edge("retrieve", "generate")
graph_builder.add_edge("generate", END)

rag_graph = graph_builder.compile()

print("LangGraph RAG successfully compiled!")
```

```python
from pydantic import BaseModel, Field
from langchain_core.documents import Document

class RAGState(BaseModel):
    question: str = ""
    documents: list[Document] = Field(default_factory=list)
    answer: str = ""
    relevant: bool = False
    sources: str = ""
    sources: list[str] = Field(default_factory=list)
```

```python

def grade_documents(state: RAGState):

    question = state["question"]
    relevant_docs = []

    for doc in state["documents"]:

        prompt = f"""
You are a document relevance grader.

Question:
{question}

Document:
{doc.page_content}

Is this document relevant to answering the question?

Reply with only YES or NO.
"""

        result = ask_llm(prompt).strip().upper()

        print("Grading result:", result[:30])

        if result.startswith("YES"):
            relevant_docs.append(doc)

    return {
        "documents": relevant_docs,
        "relevant": len(relevant_docs) > 0
    }
```

```python

def no_context(state: RAGState):

    return {
        "answer": "Sorry, mujhe provided documents mein is sawal ka relevant jawab nahi mila."
    }
```

```python

def decide_route(state: RAGState):

    if state["relevant"]:
        return "generate"

    return "no_context"
```

```python

from langgraph.graph import StateGraph, START, END

graph_builder = StateGraph(RAGState)

graph_builder.add_node("retrieve", retrieve)
graph_builder.add_node("grade_documents", grade_documents)
graph_builder.add_node("generate", generate)
graph_builder.add_node("no_context", no_context)

graph_builder.add_edge(START, "retrieve")

graph_builder.add_edge("retrieve", "grade_documents")

graph_builder.add_conditional_edges(
    "grade_documents",
    decide_route,
    {
        "generate": "generate",
        "no_context": "no_context"
    }
)

graph_builder.add_edge("generate", END)
graph_builder.add_edge("no_context", END)

rag_graph = graph_builder.compile()

print("Conditional RAG Graph compiled!")
```

```python
def rewrite_query(state: RAGState):

    question = state.question

    prompt = f"""
You are a search query optimizer.

Rewrite the following question into a clear
search query for retrieving relevant documents.

Keep the original meaning.

Original question:
{question}

Return only the rewritten query.
"""

    improved_question = ask_llm(prompt)

    return {
        "question": improved_question.strip()
    }
```

```python
def retrieve(state: RAGState):

    documents = loaded_vectorstore.similarity_search(
        state.question, k=3
    )

    return {
        "documents": documents,
        "retry_count": state.retry_count + 1
    }
```

```python
def generate(state: RAGState):

    context = "\n\n".join(
        [
            f"[Source {i+1}]\n"
            f"File: {doc.metadata.get('source', 'Unknown')}\n"
            f"Page: {doc.metadata.get('page_label', 'Unknown')}\n"
            f"Content: {doc.page_content}"
            for i, doc in enumerate(state.documents)
        ]
    )

    prompt = f"""
    Answer the question using only the context.

    Context:
    {context}

    Question:
    {state.question}

    Instructions:
    1. Cite supporting information using [1], [2], etc.
    2. Each citation must refer to the matching Source number.
    3. If context does not answer the question, say "I don't know".
    4. Never invent citations.

    Answer:
    """

    answer = ask_llm(prompt)

    sources = []

    for doc in state.documents:
        source = doc.metadata.get("source", "Unknown")
        page = doc.metadata.get("page_label", "Unknown")
        citation = f"{source}, Page {page}"

        if citation not in sources:
            sources.append(citation)

    return {
        "answer": answer,
        "sources": sources
    }
```
